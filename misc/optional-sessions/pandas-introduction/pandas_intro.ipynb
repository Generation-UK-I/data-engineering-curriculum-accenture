```python

#Pandas - Series

#A pandas *series* is a one-dimensional array of data. 

#This is quite like a simple list or dictionary structure which you will be used to in Python.

#A series can hold any type of data (integers, strings, objects etc)

#Each item in the series has an index.
```

```python
#You can create a series from a simple array

s0 = pd.Series([10, 20, 30, 40, 50])
print(s0)
```

```python

#You can create a series from an array with an index

s1 = pd.Series([1, 2, 3, 4, 5], index=["a", "b", "c", "d", "e"])
print(s1)
```

```python

#Your data does not have to be of the same type

s2 = pd.Series([1, 2, 3, 4, 5, "six"], index=["a", "b", "c", "d", "e", "f"])
print(s2)
```

```python

#You can create a series from a dictionary

s3 = pd.Series({"b": 1, "a": 0, "c": 2})
print(s3)
```

```python

#You can also pass an index when creating a series from a dictionary, 
#which will pull out the corresponding values from the dictionary

s4 = pd.Series({"b": 1, "a": 0, "c": 2, "d": 5}, index=["a", "b"])
print(s4)

print("---")

s5 = pd.Series({"b": 1, "a": 0, "c": 2, "d": 5}, index=["a", "b", "e"])
print(s5)
```

```python

#Why do we use a series over something like a list or dictionary?
#A pandas series makes it easy to start doing anlysis!

#Pandas provides lots of built-in functions for analysing our data
```

```python

#Let's make a simple series from a list

my_series = pd.Series([10, 47, 196, 88, 251])
```

```python

#We can slice the data just like we do in python
print("Get the first element")
print(my_series[0])
print()
print("Get the first three elements")
print(my_series[:3])
print()
print("Get the last two elements")
print(my_series[-2:])
print()

#You can also searh by index like a dictionary

print("Get the value for a given index 'd'")
my_series_2 = pd.Series([10, 47, 196, 88, 251], index=["a", "b", "c", "d", "e"])
print(my_series_2["d"])
print()
```

```python

#We can very simply get summary statistics such as mean, median, standard deviation

print("my series")
print(my_series)
print()

print("Get mean")
print(my_series.mean())
print()

print("Get median")
print(my_series.median())
print()

print("Get all elements above the median")
print(my_series[my_series > my_series.median()])
print()

print("Get standard deviation")
print(my_series.std())
print()

print("Get sum")
print(my_series.sum())
print()
```

```python

#We can also map over a series quite easily and do something on each element
# - use the apply function!

print("my series")
print(my_series)
print()

print("my series - squared!")
def square(x):
    return x**2
print(my_series.apply(square))
print()

print("my series - squared using a lambda!")
print(my_series.apply(lambda z: z ** 2))
print()
```

```python

# We can also do very quick and simple visualisation of our data

my_series.plot()
my_series.plot(kind="box")
```

```python

#Pandas - Data Frame

# Generally the most common pandas data structure

#A pandas *data frame* is a *a two-dimensional labelled data structure*. 
#...which you can probably more easily think of as a table.

# You can also think of each column as a series with its own index

#A data frame organises data in rows and columns, which are labelled 
#(i.e. each row has an index, and each column has a column name)

```

```python

# You can create a data frame from multiple series put together in a dictionary

series_1 = pd.Series([1, 2, 3, 4, 5])
print(series_1)
print()

series_2 = pd.Series([10, 20, 30, 40, 50])
print(series_2)
print()

d = {"one": series_1, "two": series_2}

df = pd.DataFrame(d)
print(df)
```

```python

# You can also create a data frame from a list of dictionaries

list_of_dicts = [{"a": 1, "b": 2, "c": 3}, {"a": 10, "b": 20, "c": 30}]
print(pd.DataFrame(list_of_dicts, index=["first", "second"]))
```

```python

# Or from a CSV

my_df = pd.read_csv("test.csv")
my_df
```

```python
def remove_last_char_if_sqbrack(x):
    if x[-1] == "]":
        x = x.strip("]")
    return x

my_df['name'] = my_df['name'].apply(remove_last_char_if_sqbrack)
my_df
```

```python

# You can use head() or tail() to look at the first or last few rows
# This is obviously more useful when you have many rows

#Defaults to 5 rows
print(my_df.head())
print()
print(my_df.tail())
print()

#Or you can provide the number of rows
print(my_df.head(2))
```

```python

# You can also very easily look at a specific column using the column name

print(my_df.name)
print()
print(my_df.age)
```

```python

# So you'll notice the first column of our data is an 'id' type column which auto-increments
# However, we also have a 'student_id' column, so why not just use that as the id column?

my_df.set_index('student_id', inplace=True)
my_df.head()
```

```python

# Now let's say we don't care about students' hair colour for any of our further analysis
# We are going to drop the hair colour column

my_df.drop(columns='hair_colour', inplace=True)
my_df.head()
```

```python

# Now let's say we want to split out the 'scores' column into two separate columns, say...
#'score_test_1' and 'score_test_2'

# We are going to use my_df.scores to get the scores column as a series
# We then use the str.split function which pandas gives us for a series
# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.str.split.html
# We use the 'expand=True' argument turn the 'split string' into df columns
# We assign these to two new columns
# We will then drop the original 'scores' column

my_df[['score_test_1', 'score_test_2']] = my_df.scores.str.split(",", expand=True)
my_df.drop(columns='scores', inplace=True)
my_df
```

```python

# Let's look at if there is a pattern between the students' ages and their scores in test 1

my_df.plot.scatter(x='age',y='score_test_1')
```

```python

# What's wrong with the above? Look at the y-axis!

# We generated the score for the score_test_1 from the str.split function
# So all of those figures are strings!
# matplotlib does not know how to order them!

# Let's fix that by changing the scores to numbers
# The plot will then have a properly ordered y-axis, and we can see if there is a pattern!

my_df['score_test_1'] = my_df['score_test_1'].astype(float)
my_df.plot.scatter(x='age',y='score_test_1')
```
